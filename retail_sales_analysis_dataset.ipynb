```python
import pandas as pd
```

```python
# importing dataset

df = pd.read_csv("retail_sales_analysis_dataset.csv")
df.head()
```

```python
# Check row count
df.shape[0]
```

```python
len(df)
```

```python
# check data dypes
df.dtypes
```

```python
# converting datatype of date column
df["Date"] = pd.to_datetime(df["Date"])

print(df.dtypes)
```

```python
# check missing values

df.isnull()
```

```python
#count missing values
df.isnull().sum()
```

```python
# Renaming Multiple Columns

df = df.rename(columns={
    "Date": "Date",
    "Region": "Region",
    "Product line": "Product_Line",
    "Revenue (₹000)": "Revenue",
    "Product cost (₹000)": "Product_Cost",
    "Logistics (₹000)": "Logistics",
    "Marketing (₹000)": "Marketing",
    "Other ops (₹000)": "Other_Ops",
    "Units sold": "Units_Sold"
})

print(df.columns.tolist())
```

```python
#check duplicates 
print(df.duplicated())
```

```python
#Check for negative revenue.

invalid_revenue = df[df["Revenue"] < 0]
print(invalid_revenue)
```

```python
# check for negative cost
cost_columns = ['Product_Cost', 'Logistics', 'Marketing', 'Other_Ops']

print("Negative Costs:")
print(df[(df[cost_columns] < 0).any(axis=1)])
```

```python
#check for zero revenue

print(df[df['Revenue'] == 0])
```

```python
#check unlikely units sold

print(df[df['Units_Sold'] <= 0])
```

```python
print("Regions represented:")
print(df['Region'].unique())

print("Number of regions:", df['Region'].nunique())
```

```python
print("Product lines represented:")
print(df['Product_Line'].unique())

print("Number of product lines:", df['Product_Line'].nunique())
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

df['Month'] = df['Date'].dt.to_period('M')

print("Months represented:")
print(sorted(df['Month'].dropna().unique()))

print("Number of months:", df['Month'].nunique())
```
